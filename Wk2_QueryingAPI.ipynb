```python
import requests
import json
import time
import csv

API_KEY = "5dfd45f8ff214ae9a5974e94b5af0845" # Replace with personal API key

query_names = []
last_names = []
first_names = []

with open('namus_01:2000-05:2016.csv', mode='r', newline='') as csvfile: # Reading in names from CSV file
    reader = csv.DictReader(csvfile)
    for row in reader:
        last_names.append(row['Legal Last Name'])
        first_names.append(row['Legal First Name'])
        
num_entries = len(last_names)

for i in range(num_entries):
    query = "\"" + first_names[i] + " " + last_names[i] + "\""
    query_names.append(query)

temp_queries = query_names[0:50] # Taking a sublist to avoid running out of requests

for name in temp_queries:
    
    print(f"Searching for: {name}")
    url = f"https://newsapi.org/v2/everything?q={name}&apiKey={API_KEY}"
    response = requests.get(url)
    data = response.json()

    print(f"Found {data['totalResults']} articles")
    for article in data['articles'][:5]:
        print(f"- {article['title']}")
        
    time.sleep(1)  # Wait 1 second between requests (be nice to the API!)
```
